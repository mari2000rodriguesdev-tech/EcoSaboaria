```python
a = 3
b = 5
c = 2

#Operações matemáticas
print(a + b)
print(a + b + c)
print( a - c)
print( b * a )
print( b % a) # resto da divisão
print( a  ** b) # exponeciação
print( b // c) # divisão inteira


# Operações de comparação
print( a == b) # verifica se 'a' é igual a 'b'
print( a != b) # verifica se 'a' é diferente de 'b'
print( a < b) # menor que
print( a > b) # maior que
print( a >= b) # maior igual
print( a <= b) # menor igual

# Operadores Lógicos
print( a == b and a > b) # E lógico
print(a != b or a == b) # OU lógico
print(not (a == b)) # negação ou inersão lógica
```
